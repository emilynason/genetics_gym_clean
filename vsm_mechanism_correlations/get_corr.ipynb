```python
import pandas as pd
from pandas.errors import EmptyDataError
```

```python
chromosomes = [str(i) for i in range(1,23)] + ['X']
# mechanisms 
features = ['plddt', 'rsasa', 'foldx_ddg', 'mu_snp', 'mobidb_score', 'druggability_score', 'Aromatic', 
            'Charged', 'Nonpolar_Neutral', 'Polar_Neutral', 'delta_molar_mass', 'delta_hydropathy', 
            'ss_loop_coil', 'ss_helix', 'ss_undefined', 'glycosylation', 'lipidation', 'cross_link',
            'disulfide_bond', 'acetylation', 'methylation', 'phosphorylation', 'SUMOylation', 'ubiquitination',
            'intra_chain_hydrogen_bond', 'intra_chain_nonbonded_interaction', 'intra_chain_salt_bridge', 
            'af2bind_pocket', 'all_uniprot_sites', 'compositional_bias', 'domain', 'motif', 'region',
            'repeat', 'transmembrane', 'zinc_finger', 'topological_domain','intramembrane', 
            'signal', 'propeptide','peptide', 'chain', 'transit_peptide']
```

```python
# code to make mech_vsm_gene_corr.tsv from individual chrom/mech files

results_df = pd.DataFrame()
# combine all features
for f in features:
    # combine all stats into one table for a given feature f
    for c in chromosomes:
        try:
            results_df_c = pd.read_csv(
                f'all_stats/stats_vsm_{f}_genes_chr{c}.tsv',
                sep='\t', usecols=['vsm', 'feature', 'gene', 'corr']
            )
        except EmptyDataError:
            print(f'file for feature {f} and chromosome {c} is empty, skipping')
            continue
        # except:
        #     print(f' file is either empty or does not exist for feature {f} and chromosome {c}, skipping')
        results_df = pd.concat([results_df, results_df_c], ignore_index=True)

df = (
    results_df.pivot_table(
        index=["gene", "vsm"],
        columns="feature",
        values="corr"
    )
)
df.to_csv('mech_vsm_gene_corr.tsv', sep='\t')
```
